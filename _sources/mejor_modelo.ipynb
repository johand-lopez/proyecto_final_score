# 9. Propuesta de Modelo Original

### LAVA-Embedding + XGBoost (Agregación Local de Características Correlacionadas)

**Ventajas de la implementación:**

* **Reduce la multicolinealidad:** Agrupa variables financieras (como `monthly_inhand_salary` y `monthly_balance`) que están altamente correlacionadas.

* **Captura información estructural:** En lugar de analizar 11 variables numéricas por separado, el modelo recibe 2 *features* latentes que resumen patrones complejos (p. ej., "Capacidad Financiera" y "Comportamiento Crediticio").
* **Compatible con GBDT:** Se integra directamente con XGBoost al aumentar el conjunto de datos de entrenamiento.

**Aplicación directa:**

* Se identificaron 2 "Grupos LAVA" de variables altamente correlacionadas usando la matriz de correlación.
* Se aplicó **PCA (n=1)** a cada grupo para crear 2 nuevas *features* latentes.
* Se entrenó un modelo **XGBoost** estándar usando el dataset original *aumentado* con estas 2 *features* LAVA.

**Por qué destaca:** Mejora la representación de los datos al capturar las interdependencias entre las variables, permitiendo al XGBoost identificar patrones más robustos (Resultado: 0.7821 de *F1 Score Macro* en la clasificación de tres clases).

***Referencia metodológica basada en:***
Liedl, E. R., Yassin, S. M., Kasapi, M., & Posma, J. M. (2024). Topological embedding and directional feature importance in ensemble classifiers for multi-class classification. Computational and Structural Biotechnology Journal, 23, 4108-4123. https://doi.org/10.1016/j.csbj.2024.11.013
