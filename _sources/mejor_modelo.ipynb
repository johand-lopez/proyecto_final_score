# 9. Propuesta de Modelo Original

### HAFL-Boosting (Highly Adaptive Focal Loss + Boosting)

**Ventajas para nuestro problema:**

- Diseñado específicamente para datos multi-clase desbalanceados.  
- Introduce *Highly Adaptive Focal Loss* que aumenta la atención del modelo a las clases minoritarias.  
- Compatible con los modelos que dieron mejores scores (XGBoost y Random Forest)

**Aplicación directa:** Sustituir la función de pérdida actual por HAFL o implementarla dentro de un custom objective en XGBoost.  

**Por qué destaca:** mejora F1 y *recall* en clases minoritarias sin sacrificar interpretabilidad ni velocidad.
